# Convoluciones color

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('./images/img1.jpeg', cv2.IMREAD_COLOR)

red = img[:,:,2]
green = img[:,:,1]
blue = img[:,:,0]

mascara = np.array(
  [[0, -1, 0],
   [1, 5, -1],
   [0, 1, 0]])

# Enfoque
img_enfoque = np.zeros_like(img)
img_enfoque[:,:,0] = cv2.filter2D(red, -1, mascara)
img_enfoque[:,:,1] = cv2.filter2D(green, -1, mascara)
img_enfoque[:,:,2] = cv2.filter2D(blue, -1, mascara)

# Desenfoque
kernel5x5 = np.ones((5, 5), np.float32) / 36
img_suavizado = np.zeros_like(img)
img_suavizado[:,:,0] = cv2.filter2D(red, -1, kernel5x5)
img_suavizado[:,:,1] = cv2.filter2D(green, -1, kernel5x5)
img_suavizado[:,:,2] = cv2.filter2D(blue, -1, kernel5x5)

# Realce de bordes
mascara = np.array(
  [[0, 0, 0],
   [-1, 1, 0],
   [0, 0, 0]])
img_realce_bordes = np.zeros_like(img)
img_realce_bordes[:,:,0] = cv2.filter2D(red, -1, mascara)
img_realce_bordes[:,:,1] = cv2.filter2D(green, -1, mascara)
img_realce_bordes[:,:,2] = cv2.filter2D(blue, -1, mascara)

# Repujado
mascara = np.array(
  [[-2, -1, 0],
   [-1, 1, 1],
   [0, 1, 2]])
img_repujado = np.zeros_like(img)
img_repujado[:,:,0] = cv2.filter2D(red, -1, mascara)
img_repujado[:,:,1] = cv2.filter2D(green, -1, mascara)
img_repujado[:,:,2] = cv2.filter2D(blue, -1, mascara)

# Deteccion de bordes
mascara = np.array(
  [[0, 1, 0],
   [1, -4, 1],
   [0, 1, 0]])
img_deteccion_bordes = np.zeros_like(img)
img_deteccion_bordes[:,:,0] = cv2.filter2D(red, -1, mascara)
img_deteccion_bordes[:,:,1] = cv2.filter2D(green, -1, mascara)
img_deteccion_bordes[:,:,2] = cv2.filter2D(blue, -1, mascara)

# Sobel
mascara = np.array(
  [[-1, 0, 1],
   [-2, 0, 2],
   [-1, 0, 1]])
img_sobel = np.zeros_like(img)
img_sobel[:,:,0] = cv2.filter2D(red, -1, mascara)
img_sobel[:,:,1] = cv2.filter2D(green, -1, mascara)
img_sobel[:,:,2] = cv2.filter2D(blue, -1, mascara)

# Sharpen
mascara = np.array(
  [[1, -2, 1],
   [-2, 5, -2],
   [1, -2, 1]])
img_sharpen = np.zeros_like(img)
img_sharpen[:,:,0] = cv2.filter2D(red, -1, mascara)
img_sharpen[:,:,1] = cv2.filter2D(green, -1, mascara)
img_sharpen[:,:,2] = cv2.filter2D(blue, -1, mascara)

# Norte
mascara = np.array(
  [[1, 1, 1],
   [1, -2, 1],
   [-1, -1, -1]])
img_norte = np.zeros_like(img)
img_norte[:,:,0] = cv2.filter2D(red, -1, mascara)
img_norte[:,:,1] = cv2.filter2D(green, -1, mascara)
img_norte[:,:,2] = cv2.filter2D(blue, -1, mascara)

# Este
mascara = np.array(
  [[-1, 1, 1],
   [-1, -2, 1],
   [-1, 1, 1]])
img_este = np.zeros_like(img)
img_este[:,:,0] = cv2.filter2D(red, -1, mascara)
img_este[:,:,1] = cv2.filter2D(green, -1, mascara)
img_este[:,:,2] = cv2.filter2D(blue, -1, mascara)

# Gauss
mascara = np.array(
  [
    [1, 2, 3, 1, 1],
    [2, 7, 11, 7, 2],
    [3, 11, 17, 11, 3],
    [2, 7, 11, 7, 1],
    [1, 2, 3, 2, 1]
  ])
img_gauss = np.zeros_like(img)
img_gauss[:,:,0] = cv2.filter2D(red, -1, mascara)
img_gauss[:,:,1] = cv2.filter2D(green, -1, mascara)
img_gauss[:,:,2] = cv2.filter2D(blue, -1, mascara)

plt.figure(figsize=(10, 20))

plt.subplot(6, 2, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))

plt.subplot(6, 2, 2)
plt.imshow(img_enfoque)

plt.subplot(6, 2, 3)
plt.imshow(img_suavizado)

plt.subplot(6, 2, 4)
plt.imshow(img_realce_bordes)

plt.subplot(6, 2, 5)
plt.imshow(img_repujado)

plt.subplot(6, 2, 6)
plt.imshow(img_deteccion_bordes)

plt.subplot(6, 2, 7)
plt.imshow(img_sobel)

plt.subplot(6, 2, 8)
plt.imshow(img_sharpen)

plt.subplot(6, 2, 9)
plt.imshow(img_norte)

plt.subplot(6, 2, 10)
plt.imshow(img_este)

plt.subplot(6, 2, 11)
plt.imshow(img_este)
```
