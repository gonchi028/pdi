# Convolucion horizontal

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def convolucion_media_horizontal(imagen, tamanioKernel):
  kernel = np.ones((1, tamanioKernel), np.float32) / tamanioKernel
  img_final = cv2.filter2D(imagen, -1, kernel)
  return img_final

img = cv2.imread('./images/distorsionada-horizontal.jpeg')
img_gris = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

img_resultado = convolucion_media_horizontal(img_gris, 17)

plt.figure(figsize=(15, 10))

plt.subplot(1, 2, 1)
plt.title('Imagen original')
plt.imshow(img_gris, cmap='gray')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.title('Resultado')
plt.imshow(img_resultado, cmap='gray')
plt.axis('off')
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def convolucion_media_horizontal(imagen, tamanioKernel):
  kernel = np.ones((1, tamanioKernel), np.float32) / tamanioKernel
  img_final = cv2.filter2D(imagen, -1, kernel)
  return img_final

# img = cv2.imread('./images/distorsionada-horizontal.jpeg')
img = cv2.imread('./images/distorsionada.jpg')
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

size = 25
img_result_r = convolucion_media_horizontal(img_rgb[:, :, 0], size)
img_result_g = convolucion_media_horizontal(img_rgb[:, :, 1], size)
img_result_b = convolucion_media_horizontal(img_rgb[:, :, 2], size)

img_resultado = np.stack([img_result_r, img_result_g, img_result_b], axis=2)

plt.figure(figsize=(15, 10))

plt.subplot(1, 2, 1)
plt.title('Imagen original')
plt.imshow(img_rgb)
plt.axis('off')

plt.subplot(1, 2, 2)
plt.title('Resultado')
plt.imshow(img_resultado)
plt.axis('off')
```
