# Recuperatorio segundo momento evaluativo

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
from num2words import num2words

# imagen_bgr = cv2.imread('./images/reloj.jpeg')
imagen_bgr = cv2.imread('./images/reloj-2.png')
imagen_rgb = cv2.cvtColor(imagen_bgr, cv2.COLOR_BGR2RGB)

gris = cv2.cvtColor(imagen_bgr, cv2.COLOR_BGR2GRAY)
gris = cv2.GaussianBlur(gris, (5, 5), 1)
gris = cv2.equalizeHist(gris)

bordes = cv2.Canny(gris, 50, 150)

# Detección de líneas
# lineas = cv2.HoughLinesP(bordes, 1, np.pi/180, threshold=80, minLineLength=50, maxLineGap=1)
lineas = cv2.HoughLinesP(bordes, 3, np.pi/180, threshold=50, minLineLength=30, maxLineGap=1)

# Visualizar lineas
# for linea in lineas:
#   x1, y1, x2, y2 = linea[0]
#   cv2.line(imagen_bgr, (x1, y1), (x2, y2), (0, 0, 255), 2)
# plt.imshow(imagen_bgr)
# plt.show()

centro_x = imagen_bgr.shape[1] // 2
centro_y = imagen_bgr.shape[0] // 2

def calcular_angulo(x1, y1, x2, y2):
  dist1 = np.hypot(x1 - centro_x, y1 - centro_y)
  dist2 = np.hypot(x2 - centro_x, y2 - centro_y)
  if dist1 > dist2:
    dx = x1 - centro_x
    dy = centro_y - y1
  else:
    dx = x2 - centro_x
    dy = centro_y - y2
  angulo = np.degrees(np.arctan2(dy, dx))
  angulo = (90 - angulo) % 360
  return angulo

manecillas = []
if lineas is not None:
  for linea in lineas:
    x1, y1, x2, y2 = linea[0]
    longitud = np.hypot(x2 - x1, y2 - y1)
    distancia_centro = min(np.hypot(x1 - centro_x, y1 - centro_y),np.hypot(x2 - centro_x, y2 - centro_y))
    # LINEAS CENTRO
    if distancia_centro < 100 and 40 < longitud < 300:
      angulo = calcular_angulo(x1, y1, x2, y2)
      manecillas.append({
        'puntos': (x1, y1, x2, y2),
        'longitud': longitud,
        'angulo': angulo
      })

if len(manecillas) >= 2:
  # CORTO LARGO
  manecillas.sort(key=lambda x: x['longitud'])
  horario = manecillas[0]
  minutero = manecillas[1]
  segundero = manecillas[-1]
  
  hora = int((horario['angulo'] / 30) % 12)
  minutos = int((minutero['angulo'] / 6) % 60)
  segundos = int((segundero['angulo'] / 6) % 60)

  fraccion_hora = (horario['angulo'] / 30) % 12
  if abs(fraccion_hora - hora) > 0.5:
    hora = (hora + 1) % 12

  imagen_resultado = imagen_rgb.copy()
  # cv2.circle(imagen_resultado, (centro_x, centro_y), 5, (0, 0, 255), -1)
  cv2.line(imagen_resultado,(horario['puntos'][0], horario['puntos'][1]),(horario['puntos'][2], horario['puntos'][3]), (0, 0, 255), 5)  # horario azul
  cv2.line(imagen_resultado,(minutero['puntos'][0], minutero['puntos'][1]),(minutero['puntos'][2], minutero['puntos'][3]),(0, 255, 0), 5)  # minutero verde
  cv2.line(imagen_resultado,(segundero['puntos'][0], segundero['puntos'][1]),(segundero['puntos'][2], segundero['puntos'][3]),(255, 0, 0), 5)  # segundero rojo

  hora_final = f'{hora}:{minutos:02d}:{segundos:02d}'

  hora_en_palabras = num2words(hora, lang='es')
  minutos_en_palabras = num2words(minutos, lang='es')
  segundos_en_palabras = num2words(segundos, lang='es')

  plt.figure(figsize=(10, 10))
  plt.imshow(imagen_resultado)
  plt.title(f'Hora: {hora_final}\n{hora_en_palabras} con {minutos_en_palabras} minutos y {segundos_en_palabras} segundos')
  plt.show()
```
