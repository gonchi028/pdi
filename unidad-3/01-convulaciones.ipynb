```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('./images/img1.jpeg', 0)
mascara = np.array(
  [[0, -1, 0],
   [1, 5, -1],
   [0, 1, 0]])
# mascara = np.array(
#   [[0, 0, 0],
#    [-1, 1, 0],
#    [0, 0, 0]])
img_fin = cv2.filter2D(img, -1, mascara)

plt.figure(figsize=(15, 7))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap="gray")

plt.subplot(1, 2, 2)
plt.imshow(img_fin, cmap="gray")
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('./images/img1.jpeg', 0)
kernel5x5 = np.ones((5, 5), np.float32) / 25 # suavizado
kernel7x7 = np.ones((5, 5), np.float32) / 49 # suavizado 2

suavizado = cv2.filter2D(img, -1, kernel5x5)
suavizado2 = cv2.filter2D(img, -1, kernel7x7)

plt.figure(figsize=(15, 7))

plt.subplot(2, 2, 1)
plt.title("Imagen original")
plt.imshow(img, cmap="gray")
plt.axis('off')

plt.subplot(2, 2, 2)
plt.title("Imagen suavizada")
plt.imshow(suavizado, cmap="gray")
plt.axis('off')

plt.subplot(2, 2, 3)
plt.title("Imagen suavizada 2")
plt.imshow(suavizado, cmap="gray")
plt.axis('off')
```
