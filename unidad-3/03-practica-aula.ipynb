# Aliasing y los patrones de moire

El aliasing de una imagen es un fenómeno visual en la imagen digital donde una representación muestreada presenta artefactos o distorsiones indeseables. Estos artefactos suelen aparecer como bordes irregulares (jagged) o escalonados ("stair-stepped"), lo que puede reducir significativamente la calidad y el detalle general de la imagen. La causa principal del aliasing de imagen son las limitaciones de los sensores o pantallas digitales, que se componen de píxeles discretos. Cuando una imagen se captura o se muestra con un número limitado de píxeles, el contenido de alta frecuencia (como líneas finas, texturas o patrones intrincados) puede resultar difícil de resolver, lo que resulta en aliasing.

El aliasing no se limita a un único artefacto visual; también incluye efectos como patrones de Moiré (patrones extraños en forma de ondas o remolinos) y ruido de color (cambios de color aleatorios). Además, el aliasing puede ocurrir no solo en el espacio (aliasing espacial) sino también en el tiempo (aliasing temporal), como en vídeos o animaciones, donde el movimiento se ve brusco o poco natural. Estos problemas surgen debido a la interpretación errónea de frecuencias superiores a la mitad de la frecuencia de muestreo, lo que provoca que se perciban como frecuencias más bajas. Este fenómeno, resultado del submuestreo, puede hacer que las señales o los detalles sean indistinguibles, lo que genera errores de medición en las imágenes digitales.

### Patron de Moiré

El efecto moiré es un patrón de interferencia que se forma cuando se superponen dos rejillas de líneas, ya sean rectas o curvas, con un cierto ángulo. O cuando las rejillas tienen tamaños ligeramente diferentes.

# Contrarestando los efectos

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt


img1 = cv2.imread('./images/img-p1.jpeg', 0)
img2 = cv2.imread('./images/img-p2.jpeg', 0)
img3 = cv2.imread('./images/img-p3.jpeg', 0)

def convolucion_media_horizontal(imagen, tamanioKernel):
  kernel = np.ones((1, tamanioKernel), np.float32) / tamanioKernel
  img_final = cv2.filter2D(imagen, -1, kernel)
  return img_final

def convolucion_media_vertical(imagen, tamanioKernel):
  kernel = np.ones((tamanioKernel, 1), np.float32) / tamanioKernel
  img_final = cv2.filter2D(imagen, -1, kernel)
  return img_final

def convolucion_media_diagonal_1(imagen, tamanioKernel):
  kernel = np.eye(tamanioKernel) * np.ones((tamanioKernel, tamanioKernel)) / tamanioKernel
  img_final = cv2.filter2D(imagen, -1, kernel)
  return img_final
def convolucion_media_diagonal_2(imagen, tamanioKernel):
  kernel = np.zeros((tamanioKernel, tamanioKernel), np.float32)
  for i in range(tamanioKernel):
    kernel[i][tamanioKernel - i - 1] = 1
  kernel = kernel / tamanioKernel
  img_final = cv2.filter2D(imagen, -1, kernel)
  return img_final

def enfoque(img):
  mascara = np.array(
  [[0, -1, 0],
   [-1, 5, -1],
   [0, -1, 0]])
  mascara = mascara
  img_final = cv2.filter2D(img, -1, mascara)
  return img_final

img_resultado_1 = convolucion_media_vertical(img1, 9)
img_resultado_1 = convolucion_media_horizontal(img_resultado_1, 9)
img_resultado_1 = enfoque(img_resultado_1)

img_resultado_2 = convolucion_media_diagonal_1(img2, 9)
img_resultado_2 = convolucion_media_diagonal_2(img_resultado_2, 9)

img_resultado_3 = convolucion_media_vertical(img3, 9)
img_resultado_3 = convolucion_media_horizontal(img_resultado_3, 9)
img_resultado_3 = enfoque(img_resultado_3)

plt.figure(figsize=(10, 15))

plt.subplot(3, 2, 1)
plt.title('Imagen original')
plt.imshow(img1, cmap='gray')
plt.axis('off')

plt.subplot(3, 2, 2)
plt.title('Resultado')
plt.imshow(img_resultado_1, cmap='gray')
plt.axis('off')

plt.subplot(3, 2, 3)
plt.title('Imagen original')
plt.imshow(img2, cmap='gray')
plt.axis('off')

plt.subplot(3, 2, 4)
plt.title('Resultado')
plt.imshow(img_resultado_2, cmap='gray')
plt.axis('off')

plt.subplot(3, 2, 5)
plt.title('Imagen original')
plt.imshow(img3, cmap='gray')
plt.axis('off')

plt.subplot(3, 2, 6)
plt.title('Resultado')
plt.imshow(img_resultado_3, cmap='gray')
plt.axis('off')
```
