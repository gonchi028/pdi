# Histogramas

### Ecualizacion de histograma

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

imagenColor = cv2.imread('./images/1.jpeg', 0)

histograma, calc = np.histogram(imagenColor.flatten(), 256, [0, 256])
distAcumulada = histograma.cumsum()
distAcumladaNormalizada = distAcumulada * histograma.max() / distAcumulada.max()

ecualizado = cv2.equalizeHist(imagenColor)
histogramaEcualizado, _ = np.histogram(ecualizado.flatten(), 256, [0, 256])

plt.figure(figsize=(15, 7))

plt.subplot(2, 2, 1)
plt.title('Imagen original')
plt.imshow(imagenColor, cmap='gray')
plt.axis('off')

plt.subplot(2, 2, 2)
plt.title('Histograma original')
plt.plot(distAcumulada, color='b')
plt.hist(imagenColor.flatten(), 256, [0, 256], color='r', alpha=0.5)
plt.xlim([0, 256])
plt.legend(['Distribucion', 'histograma'], loc='upper left')

plt.subplot(2, 2, 3)
plt.title('Imagen ecualizada')
plt.imshow(ecualizado, cmap='gray')
plt.axis('off')

plt.subplot(2, 2, 4)
plt.title('Histograma ecualizado')
plt.plot(histogramaEcualizado, color='b')
plt.xlim([0, 256])

plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

imagenColor = cv2.imread('./images/1.jpeg')
imagenGris = cv2.imread('./images/1.jpeg', 0)

ecualizada = cv2.equalizeHist(imagenGris)
ecu_color = cv2.merge([cv2.equalizeHist(imagenColor[:, :, 0]), cv2.equalizeHist(imagenColor[:, :, 1]), cv2.equalizeHist(imagenColor[:, :, 2])])

plt.imshow(ecualizada, cmap='gray')
plt.show()

plt.imshow(cv2.cvtColor(ecu_color, cv2.COLOR_BGR2RGB))
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread('./images/1.jpeg', 0)

transformada = np.fft.fft2(image) # Se aplica transformada e inversa de forier
desplazar = np.fft.fftshift(transformada) # Se separa el espectro
magnitud_espectro = np.log(1 + np.abs(desplazar))

plt.imshow(image, cmap='gray')
plt.show()

plt.imshow(magnitud_espectro, cmap='gray')
plt.show()
```

# Rotar imagen

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import imutils

image = cv2.imread('./images/1.jpeg')

plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.show()

angulo = 45
rotar = imutils.rotate(image, angulo)

plt.imshow(rotar)
plt.show()
```
