# Conversiones de imagenes
Manera manual mediante ecuaciones matematicas

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

img_rgb = cv2.cvtColor(cv2.imread('./images/1.jpeg', cv2.IMREAD_COLOR), cv2.COLOR_BGR2RGB)

img_hsv = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2HSV)

img_rgb2 = cv2.cvtColor(img_hsv, cv2.COLOR_HSV2RGB)

fig, axs = plt.subplots(1, 3, figsize=(20, 10))

axs[0].imshow(img_rgb)
axs[0].set_title('Imagen original')
axs[0].axis('off')

axs[1].imshow(img_hsv, cmap='hsv')
axs[1].set_title('Imagen hsv')
axs[1].axis('off')

axs[2].imshow(img_rgb2)
axs[2].set_title('Imagen rgb')
axs[2].axis('off')
```

```python
import cv2
import matplotlib.pyplot as plt

imagen = cv2.imread('./images/1.jpeg', 0)

_, umbralBinario = cv2.threshold(imagen, 127, 255, cv2.THRESH_BINARY)
_, umbralBinarioInv = cv2.threshold(imagen, 127, 255, cv2.THRESH_BINARY_INV)

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(imagen, cmap='gray')
plt.title('Imagen original')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(umbralBinario, cmap='gray')
plt.title('Umbral binario')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(umbralBinarioInv, cmap='gray')
plt.title('Umbral binario inverso')
plt.axis('off')
```
