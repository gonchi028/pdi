# Solarizacion

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def solarizar(imagen, umbral):
  imagen_solarizada = np.where(imagen < umbral, imagen, 255 - imagen)
  return imagen_solarizada

imagen = cv2.imread('./images/1.jpeg')
img_gris = cv2.cvtColor(imagen, cv2.COLOR_BGR2GRAY)
umbral = 127
img_solarizada = solarizar(img_gris, umbral)

plt.figure(figsize=(10, 5))
plt.subplot(121)
plt.imshow(cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB), cmap='gray')
plt.title('Imagen Original')
plt.axis('off')

plt.subplot(122)
plt.imshow(img_solarizada, cmap='gray')
plt.title('Imagen Solarizada')
plt.axis('off')
```

# Posterizada

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def posterizar(imagen, nivel):
  factor = 255 / nivel
  imagen_posterizada = np.round((imagen / factor) * factor)
  return imagen_posterizada.astype(np.uint8)

imagen = cv2.imread('./images/1.jpeg')
imag_gris = cv2.cvtColor(imagen, cv2.COLOR_BGR2GRAY)
nivel = 4

img_posterizada = posterizar(imag_gris, nivel)

plt.figure(figsize=(10, 5))
plt.subplot(121)
plt.imshow(cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB), cmap='gray')
plt.title('Imagen Original')
plt.axis('off')

plt.subplot(122)
plt.imshow(img_posterizada, cmap='gray')
plt.title('Imagen Posterizada')
plt.axis('off')
```

# Falso color

Para ver detalles que no se pueden ver a simple vista

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def falso_color(imagen_gray):
  escala_imagen = cv2.normalize(imagen_gray, None, 0, 255, cv2.NORM_MINMAX)
  
  aplio_color = cv2.applyColorMap(escala_imagen, cv2.COLORMAP_INFERNO)
  return aplio_color

imagen_gray = cv2.imread('./images/1.jpeg', 0)

img_falso_color = falso_color(imagen_gray)

plt.figure(figsize=(10, 5))
plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(imagen_gray, cv2.COLOR_BGR2RGB), cmap='gray')
plt.title('Imagen Original')
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(cv2.cvtColor(img_falso_color, cv2.COLOR_BGR2RGB), cmap='gray')
plt.title('Imagen Falso Color')
plt.axis('off')
```
