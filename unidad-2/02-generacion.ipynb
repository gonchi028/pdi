# Creacion de una imagen

```python
import matplotlib.pyplot as plt
import cv2
import numpy as np

plt.rcParams['figure.figsize'] = [10, 5]
imgSize = 100

R = np.ones((imgSize, imgSize)) * 255
G = np.ones((imgSize, imgSize)) * 128
B = np.ones((imgSize, imgSize)) * 0

posX = 50
posY = 50

R[posX, posY] = 255
G[posX, posY] = 255
B[posX, posY] = 255

img = np.dstack((R, G, B))
img = np.array(img, dtype=int)

plt.figure()
plt.imshow(img)

cv2.imwrite('images/generada.png', img)
```

# Escala de grises y profundidad de color

```python
import matplotlib.pyplot as plt
import cv2
import numpy as np

img = cv2.imread('./images/1.jpeg', cv2.IMREAD_GRAYSCALE)

# Valor de profundidad de color inicial
img_inicial = img.dtype
print('Profundidad de color: ', img_inicial)

# Cambio a 16 bits
img_16bit = np.uint16(img)
img_nueva = img_16bit.dtype
print('Profundidad de color: ', img_nueva)

# Mostrando las imagenes
fig, axs = plt.subplots(1, 2, figsize=(20, 10))

axs[0].imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
axs[0].set_title('Imagen Original')
axs[0].axis('off')

axs[1].imshow(cv2.cvtColor(img_16bit, cv2.COLOR_BGR2RGB))
axs[1].set_title('Imagen 16 bits')
axs[1].axis('off')
```

# RGB a HSV

```python
import matplotlib.pyplot as plt
import cv2
import numpy as np

fig, axs = plt.subplots(1, 4, figsize=(20, 10))

img_bgr = cv2.imread('./images/1.jpeg')
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

hsv = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2HSV)

axs[0].imshow(img_rgb)
axs[0].set_title('Imagen RGB')
axs[0].axis('off')

# vmax son valores estandar
fig_img1 = axs[1].imshow(hsv[..., 0], cmap='gray', vmin = 0, vmax = 169)
fig.colorbar(fig_img1, ax=axs[1], shrink=0.3)
axs[1].set_title('canal H')

fig_img2 = axs[2].imshow(hsv[..., 1], cmap='gray', vmin = 0, vmax = 255)
fig.colorbar(fig_img2, ax=axs[2], shrink=0.3)
axs[2].set_title('canal S')

fig_img3 = axs[3].imshow(hsv[..., 2], cmap='gray', vmin = 0, vmax = 255)
fig.colorbar(fig_img3, ax=axs[3], shrink=0.3)
axs[3].set_title('canal V')
```
