# Umbralizaciones por defecto tuzero en opencv

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('./images/1.jpeg', 0) # Lectura de imagen en escala de grises

threshold_value = 127 # Valor de umbral

_, umbral_trunc = cv2.threshold(img, threshold_value, 255, cv2.THRESH_TRUNC) # Umbral truncado
_, umbral_tozero = cv2.threshold(img, threshold_value, 255, cv2.THRESH_TOZERO) # Umbral a cero
_, umbral_tozero_inv = cv2.threshold(img, threshold_value, 255, cv2.THRESH_TOZERO_INV) # Umbral a cero inverso

plt.figure(figsize=(15, 7))

plt.subplot(2,2,1)
plt.imshow(img, cmap='gray')
plt.title('Imagen Original')
plt.axis('off')

plt.subplot(2,2,2)
plt.imshow(umbral_trunc, cmap='gray')
plt.title('Umbral Truncado')
plt.axis('off')

plt.subplot(2,2,3)
plt.imshow(umbral_tozero, cmap='gray')
plt.title('Umbral a Cero')
plt.axis('off')

plt.subplot(2,2,4)
plt.imshow(umbral_tozero_inv, cmap='gray')
plt.title('Umbral a Cero Inverso')
plt.axis('off')

plt.show()
```

# Contraste

Sirve para reconocer objetos, donde no haya habido tanta luz

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def constraste(img, factor = 1.0):
  img = img.astype(np.float32) # Convertir a float32 para evitar desbordamiento
  media = np.mean(img) # Calcular la media de la imagen
  img_constraste = media + (img - media) * factor # Aplicar el factor de contraste
  img_constraste = np.clip(img_constraste, 0, 255) # Limitar los valores entre 0 y 255
  img_constraste = img_constraste.astype(np.uint8) # Convertir de nuevo a uint8
  return img_constraste

img = cv2.imread('./images/1.jpeg', 0) # Lectura de imagen en escala de grises
factor = 4.0 # Factor de contraste
img_contraste = constraste(img, factor) # Aumentar el contraste

plt.figure(figsize=(15, 7))
plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title('Imagen Original')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(img_contraste, cmap='gray')
plt.title('Imagen con Contraste Aumentado')
plt.axis('off')

plt.show()
```

### Contraste a color

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def contraste_color(img, factor = 1.0):
  img = img.astype(np.float32) # Convertir a float32 para evitar desbordamiento
  for i in range(3):
    media = np.mean(img[:, :, i])
    img[:, :, i] = media + (img[:, :, i] - media) * factor
    img[:, :, i] = np.clip(img[:, :, i], 0, 255)
  return img.astype(np.uint8)

img = cv2.imread('./images/1.jpeg') # Lectura de imagen en color
factor = 5.0 # Factor de contraste
img_contraste = contraste_color(img, factor) # Aumentar el contraste

plt.figure(figsize=(15, 7))
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Imagen Original')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(img_contraste, cv2.COLOR_BGR2RGB))
plt.title('Imagen con Contraste Aumentado')
plt.axis('off')

plt.show()
```

# Contraste y brillo

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def ajuste(img, contraste = 1.0, brillo = 0):
  img = img.astype(np.float32) # Convertir a float32 para evitar desbordamiento
  media = np.mean(img) # Calcular la media de la imagen
  img_ajustada = media + (img - media) * contraste + brillo # Aplicar el factor de contraste y brillo
  img_ajustada = np.clip(img_ajustada, 0, 255) # Limitar los valores entre 0 y 255
  img_ajustada = img_ajustada.astype(np.uint8) # Convertir de nuevo a uint8
  return img_ajustada

img = cv2.imread('./images/1.jpeg')

valor_contraste = 1.5 # Valor de contraste
valor_brillo = 20 # Valor de brillo
img_nueva = ajuste(img, valor_contraste, valor_brillo) # Aumentar el contraste y brillo

# Trabajamos en bgr pero para mostrar la imagen en matplotlib la convertimos a rgb

img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
img_nueva_rgb = cv2.cvtColor(img_nueva, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(15, 7))

plt.subplot(1, 2, 1)
plt.imshow(img_rgb)
plt.title('Imagen Original')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(img_nueva_rgb)
plt.title('Imagen Ajustada')
plt.axis('off')

plt.show()
```

# Histograma de una imagen contrastada

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def contraste(img, factor = 2.0):
  img = img.astype(np.float32) # Convertir a float32 para evitar desbordamiento
  media = np.mean(img) # Calcular la media de la imagen
  img_contraste = media + (img - media) * factor # Aplicar el factor de contraste
  img_contraste = np.clip(img_contraste, 0, 255) # Limitar los valores entre 0 y 255
  img_contraste = img_contraste.astype(np.uint8) # Convertir de nuevo a uint8
  return img_contraste

img = cv2.imread('./images/1.jpeg', 0) # Lectura de imagen en escala de grises
factor = 6.0 # Factor de contraste

img_nueva = contraste(img, factor) # Aumentar el contraste

histograma_original = cv2.calcHist([img], [0], None, [256], [0, 256]) # Histograma de la imagen original
histograma_nueva = cv2.calcHist([img_nueva], [0], None, [256], [0, 256]) # Histograma de la imagen ajustada

plt.figure(figsize = (15, 12))

plt.subplot(2, 2, 1)
plt.imshow(img, cmap='gray')
plt.title('Imagen Original')
plt.axis('off')

plt.subplot(2, 2, 2)
plt.plot(histograma_original, color='red')
plt.title('Histograma Original')
plt.xlabel('Intensidad')
plt.ylabel('Frecuencia')

plt.subplot(2, 2, 3)
plt.imshow(img_nueva, cmap='gray')
plt.title('Imagen Ajustada')
plt.axis('off')

plt.subplot(2, 2, 4)
plt.plot(histograma_nueva)
plt.title('Histograma Ajustado')
plt.xlabel('Intensidad')
plt.ylabel('Frecuencia')

plt.show()
```

# Ejercicio 1

Lectura, punto de interes e histograma en 3 canales

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['figure.figsize'] = (15, 7) # Tamaño de la figura

imagen = cv2.imread('./images/1.jpeg') # Lectura de imagen
imagen_rgb = cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB) # Convertir de BGR a RGB

# Mostrando una imagen
plt.imshow(imagen_rgb) # Mostrar imagen
plt.show()

# Seleccionando punto de interes
plt.imshow(imagen_rgb[300:500, 600:1000, :]) # Mostrar imagen
plt.show()

# Guardar la imagen seccionada
img_cortada = imagen_rgb[300:500, 600:1000, :]
plt.imsave('./images/imagen_cortada.jpeg', img_cortada) # Guardar imagen

# Histograma de canales de la imagen seccionada
seccionada = plt.imread('./images/imagen_cortada.jpeg') # Lectura rgb
color = ('r', 'g', 'b')

for canal, col in enumerate(color):
  histograma = cv2.calcHist([seccionada * 255], [canal], None, [256], [0, 256]) # Histograma
  plt.plot(histograma, color=col) # Graficar histograma
  plt.xlim([0, 256]) # Limitar el eje x

plt.show()
```

# Mascaras

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

imagen = cv2.imread('./images/1.jpeg', 0)

alto, ancho = imagen.shape # Obtener dimensiones de la imagen

# Crear una mascara circular
mascara = np.zeros_like(imagen) 
centro = (ancho // 2, alto // 2) # Centro de la mascara

radio = 200 # Radio de la mascara
cv2.circle(mascara, centro, radio, (255), -1) # Crear mascara circular

imagen_mascara = cv2.bitwise_and(imagen, imagen, mask=mascara) # Aplicar mascara a la imagen

plt.figure(figsize=(15, 7))

plt.subplot(1, 2, 1)
plt.imshow(imagen, cmap='gray')
plt.title('Imagen Original')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(imagen_mascara, cmap='gray')
plt.title('Imagen con Mascara Circular')
plt.axis('off')

plt.show()
```
