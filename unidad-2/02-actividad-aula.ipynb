# Actividad Aula

```python
import matplotlib.pyplot as plt
import cv2
import numpy as np

plt.rcParams['figure.figsize'] = [10, 5]
imgSize = 100

R = np.ones((imgSize, imgSize)) * 255
G = np.ones((imgSize, imgSize)) * 128
B = np.ones((imgSize, imgSize)) * 0

for i in range(100):
  for j in range(100):
    if (i < 50) :
      if (j < 50):
        R[i, j] = 255
        G[i, j] = 0
        B[i, j] = 0
      else:
        R[i, j] = 0
        G[i, j] = 255
        B[i, j] = 0
    else:
      if (j < 50):
        R[i, j] = 0
        G[i, j] = 0
        B[i, j] = 255
      else:
        R[i, j] = 0
        G[i, j] = 0
        B[i, j] = 0

img = np.dstack((R, G, B))
img = np.array(img, dtype=int)

plt.figure()
plt.imshow(img)

cv2.imwrite('images/activdad-aula-generado.png', img)
```
