# Ejercicio 7

Desarrolla una aplicación que ordene números a través del método burbuja.

```python
def ordenarBurbuja(arreglo):
  n = len(arreglo)
  for i in range(n - 1):
    for j in range(n - 1 - i):
      if arreglo[j] > arreglo[j+1]:
        arreglo[j], arreglo[j+1] = arreglo[j+1], arreglo[j]

arreglo = [64, 34, 25, 12, 22, 11, 90]
print(f"Arreglo original: {arreglo}")
ordenarBurbuja(arreglo)
print(f"Arreglo ordenado: {arreglo}")
```
