# Actividad de laboratorio
Procesando imagenes

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

# Carga de las 3 imagenes
img1_bgr = cv2.imread('images/1.jpeg')
img2_bgr = cv2.imread('images/2.jpeg')
img3_bgr = cv2.imread('images/3.jpeg')

# Convertir a rgb
img1_rgb = cv2.cvtColor(img1_bgr, cv2.COLOR_BGR2RGB)
img2_rgb = cv2.cvtColor(img2_bgr, cv2.COLOR_BGR2RGB)
img3_rgb = cv2.cvtColor(img3_bgr, cv2.COLOR_BGR2RGB)

# Mostrar las 3 imagenes
fig,axs = plt.subplots(1, 3, figsize=(20, 5))

axs[0].imshow(img1_rgb)
axs[0].axis('off')
axs[0].set_title('Imagen 1')

axs[1].imshow(img2_rgb)
axs[1].axis('off')
axs[1].set_title('Imagen 2')

axs[2].imshow(img3_rgb)
axs[2].axis('off')
axs[2].set_title('Imagen 3')

# Convertir imagen a escala de grises
img1_grises = cv2.cvtColor(img1_bgr, cv2.COLOR_BGR2GRAY)
img2_grises = cv2.cvtColor(img2_bgr, cv2.COLOR_BGR2GRAY)
img3_grises = cv2.cvtColor(img3_bgr, cv2.COLOR_BGR2GRAY)

# Guardar imagenes en escala de grises
cv2.imwrite('images/grises_1.jpeg', img1_grises)
cv2.imwrite('images/grises_2.jpeg', img2_grises)
cv2.imwrite('images/grises_3.jpeg', img3_grises)

# Guardar imagenes bgr
plt.imsave('images/bgr_1.jpeg', img1_bgr)
plt.imsave('images/bgr_2.jpeg', img2_bgr)
plt.imsave('images/bgr_3.jpeg', img3_bgr)

# Mostrar las 3 imagenes en escala de grises
fig,axs = plt.subplots(1, 3, figsize=(20, 5))

axs[0].imshow(img1_grises, cmap='gray')
axs[0].axis('off')
axs[0].set_title('Imagen 1')

axs[1].imshow(img2_grises, cmap='gray')
axs[1].axis('off')
axs[1].set_title('Imagen 2')

axs[2].imshow(img3_grises, cmap='gray')
axs[2].axis('off')
axs[2].set_title('Imagen 3')
```

# Mostrando las imagenes

```python
# Funciones para mostrar imagenes
def mostrar_imagen(img):
  # Escala de grises por canales
  plt.rcParams['figure.figsize'] = [10, 5]

  plt.subplot(3, 4, 1)
  plt.axis('off')
  plt.title('Imagen Original')
  plt.imshow(img)

  plt.subplot(3, 4, 2)
  plt.axis('off')
  plt.title('Canal Rojo')
  plt.imshow(img[:,:,0], cmap='gray')

  plt.subplot(3, 4, 3)
  plt.axis('off')
  plt.title('Canal Verde')
  plt.imshow(img[:,:,1], cmap='gray')

  plt.subplot(3, 4, 4)
  plt.axis('off')
  plt.title('Canal Azul')
  plt.imshow(img[:,:,2], cmap='gray')

  # Colores reales
  
  red = img[:,:,0]
  green = img[:,:,1]
  blue = img[:,:,2]

  red_image = np.zeros_like(img)
  red_image[:,:,0] = red
  green_image = np.zeros_like(img)
  green_image[:,:,1] = green
  blue_image = np.zeros_like(img)
  blue_image[:,:,2] = blue


  plt.subplot(3, 4, 5)
  plt.axis('off')
  plt.title('Imagen Original')
  plt.imshow(img)

  plt.subplot(3, 4, 6)
  plt.axis('off')
  plt.title('Canal Rojo')
  plt.imshow(red_image)

  plt.subplot(3, 4, 7)
  plt.axis('off')
  plt.title('Canal Verde')
  plt.imshow(green_image)

  plt.subplot(3, 4, 8)
  plt.axis('off')
  plt.title('Canal Azul')
  plt.imshow(blue_image)
  
  # Colores falsos

  r, g, b = cv2.split(img)
  
  plt.subplot(3, 4, 9)
  plt.axis('off')
  plt.title('Imagen Original')
  plt.imshow(img)

  plt.subplot(3, 4, 10)
  plt.axis('off')
  plt.title('Canal Rojo')
  plt.imshow(r, cmap='Reds')

  plt.subplot(3, 4, 11)
  plt.axis('off')
  plt.title('Canal Verde')
  plt.imshow(g, cmap='Greens')

  plt.subplot(3, 4, 12)
  plt.axis('off')
  plt.title('Canal Azul')
  plt.imshow(b, cmap='Blues')

# Imagen 1
mostrar_imagen(img1_rgb)
```

```python
# Imagen 2
mostrar_imagen(img2_rgb)
```

```python
# Imagen 3
mostrar_imagen(img3_rgb)
```
