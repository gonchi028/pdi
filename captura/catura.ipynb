```python
import cv2
import matplotlib.pyplot as plt

plt.rcParams['figure.figsize'] = [10, 5]

img = cv2.imread('images/2.png', cv2.IMREAD_COLOR)

plt.subplot(1, 4, 1)
plt.title('Imagen Original')
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))

plt.subplot(1, 4, 2)
plt.title('Canal Rojo')
plt.imshow(img[:,:,0], cmap='gray')

plt.subplot(1, 4, 3)
plt.title('Canal Verde')
plt.imshow(img[:,:,1], cmap='gray')

plt.subplot(1, 4, 4)
plt.title('Canal Azul')
plt.imshow(img[:,:,2], cmap='gray')
```

# Canales y colores reales

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams['figure.figsize'] = [10, 5]

img = cv2.imread('images/2.png', cv2.IMREAD_COLOR)
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

red = img[:,:,2]
green = img[:,:,1]
blue = img[:,:,0]

red_image = np.zeros_like(img_rgb)
red_image[:,:,0] = red

green_image = np.zeros_like(img_rgb)
green_image[:,:,1] = green

blue_image = np.zeros_like(img_rgb)
blue_image[:,:,2] = blue

plt.subplot(1, 4, 1)
plt.axis('off')
plt.title('Imagen Original')
plt.imshow(img_rgb)

plt.subplot(1, 4, 2)
plt.axis('off')
plt.title('Canal Rojo')
plt.imshow(red_image)

plt.subplot(1, 4, 3)
plt.axis('off')
plt.title('Canal Verde')
plt.imshow(green_image)

plt.subplot(1, 4, 4)
plt.axis('off')
plt.title('Canal Azul')
plt.imshow(blue_image)
```

# Colores falsos

```python
import cv2
import matplotlib.pyplot as plt

plt.rcParams['figure.figsize'] = [10, 5]

img = cv2.imread('images/2.png', cv2.IMREAD_COLOR)

b, g, r = cv2.split(img)

plt.subplot(1, 4, 1)
plt.axis('off')
plt.title('Imagen Original')
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))

plt.subplot(1, 4, 2)
plt.axis('off')
plt.title('Canal Rojo')
plt.imshow(r, cmap='Reds')

plt.subplot(1, 4, 3)
plt.axis('off')
plt.title('Canal Verde')
plt.imshow(g, cmap='Greens')

plt.subplot(1, 4, 4)
plt.axis('off')
plt.title('Canal Azul')
plt.imshow(b, cmap='Blues')
```
