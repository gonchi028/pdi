```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

image = cv2.imread('./images/elementos.jpg')
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

_, binario = cv2.threshold(gray, 128, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)
borde, _ = cv2.findContours(binario, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
cont = len(borde)

plt.figure(figsize=(15, 7))

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title('Imagen original')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(binario, cmap='gray')
plt.title(f'Imagen binaria - {cont} frutas')
plt.axis('off')

plt.tight_layout()
plt.show()
```
